```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python

arquivo = 'z0019_1.csv'
caminho = 'C:\PythonEnginner\desenvolvimento\myproject\landing\z0019_1.csv'
data_ingestion = datetime.now().strftime("%d-%m-%y %H:%M:%S")
df = pd.read_csv(caminho,sep=';', encoding='latin-1', low_memory=False)
df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestion
df.head()
```

```python
con.execute("""
    CREATE TABLE IF NOT EXISTS bronze_produtos (
            NATBR VARCHAR,
            MAKTX VARCHAR,
            WERKS VARCHAR,
            MAINS VARCHAR,
            LABST VARCHAR,
            nome_arquivo VARCHAR,
            data_ingestao TIMESTAMP 
        )
""")
```

```python
#insert data into the table
con.execute("INSERT INTO bronze_produtos SELECT * FROM df")
```

```python
result = con.execute("SELECT * FROM bronze_produtos").fetchdf()
print(result)
```
